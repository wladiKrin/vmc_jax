```julia
# using Plots
using DelimitedFiles

function analyze_logging_files(folder_path::String)
    """
    Find all files starting with 'logging_' in the specified folder,
    extract timestamps from filenames and values from files, then plot.
    
    Args:
        folder_path: Path to the folder containing logging files
    
    Returns:
        Plot showing extracted values vs time
    """
    
    # Get all files in the folder that start with 'logging_'
    all_files = readdir(folder_path)
    logging_files = filter(f -> startswith(f, "logging_"), all_files)
    
    if isempty(logging_files)
        println("No files starting with 'logging_' found in $folder_path")
        return nothing
    end
    
    println("Found $(length(logging_files)) logging files")
    
    # Arrays to store extracted data
    timestamps = Float64[]
    values = Float64[]
    filenames = String[]
    
    for filename in logging_files
        try
            # Extract timestamp from filename
            # Remove 'logging_' prefix and extract floating point number
            time_str = replace(filename, "logging_" => "")
            
            # Handle case where there might be file extension
            if contains(time_str, ".")
                # Split by dots and find the floating point number
                parts = split(time_str, ".")
                if length(parts) >= 2
                    # Try to reconstruct the floating point number
                    # Assume format like "logging_123.456.txt" or "logging_123.456"
                    potential_number = parts[1] * "." * parts[2]
                    timestamp = parse(Float64, potential_number)
                else
                    timestamp = parse(Float64, parts[1])
                end
            else
                timestamp = parse(Float64, time_str)
            end
            
            # Read the file and extract value from second to last row
            filepath = joinpath(folder_path, filename)
            
            # Try different approaches to read the file
            lines = readlines(filepath)
            
            if length(lines) < 2
                println("Warning: File $filename has less than 2 lines, skipping...")
                continue
            end
            
            # Get second to last row
            second_last_row = lines[end-1]
            
            # Split the row and get the last entry (floating point number)
            # Handle different possible delimiters
            row_parts = split(strip(second_last_row))
            
            if isempty(row_parts)
                println("Warning: Second to last row in $filename is empty, skipping...")
                continue
            end
            
            # Extract the last number from the row
            last_entry = row_parts[end]
            value = parse(Float64, last_entry)
            
            # Store the extracted data
            push!(timestamps, timestamp)
            push!(values, value)
            push!(filenames, filename)
            
            # println("Processed $filename: time=$timestamp, value=$value")
            
        catch e
            println("Error processing file $filename: $e")
            continue
        end
    end
    
    if isempty(timestamps)
        println("No valid data extracted from any files")
        return nothing
    end
    
    # Sort data by timestamp
    sorted_indices = sortperm(timestamps)
    sorted_times = timestamps[sorted_indices]
    sorted_values = values[sorted_indices]
    sorted_filenames = filenames[sorted_indices]
    
    return (times=sorted_times, values=sorted_values, filenames=sorted_filenames)
end
```

```julia
function z_to_x_basis(psi_z::Vector{<:Number})
    N = length(psi_z)
    n = log2(N)
    !isinteger(n) && error("State vector length must be a power of 2")
    n = Int(n)
    psi_x = copy(psi_z)
    inv_sqrt2 = 1 / sqrt(2)

    for j in 0:(n-1)           # Iterate over each qubit
        step = 2^(j+1)         # Stride between blocks
        for base in 1:step:N    # Iterate over each block
            # Process half of the block (size 2^j)
            for i in 0:(2^j - 1)
                idx0 = base + i
                idx1 = idx0 + 2^j
                a = psi_x[idx0]
                b = psi_x[idx1]
                # Apply Hadamard transformation
                psi_x[idx0] = (a + b) * inv_sqrt2
                psi_x[idx1] = (a - b) * inv_sqrt2
            end
        end
    end
    return psi_x
end
```
