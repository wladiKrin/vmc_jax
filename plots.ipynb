```julia
import jax
import jax.numpy as jnp
from jax import random
from typing import Tuple, Optional


def randomized_eigh(
    A: jnp.ndarray,
    k: int,
    n_oversamples: int = 10,
    n_iter: int = 2,
    key: Optional[jax.random.PRNGKey] = None
) -> Tuple[jnp.ndarray, jnp.ndarray]:
    """
    Compute a randomized eigendecomposition of symmetric matrix A.
    
    This finds the top-k eigenvalues and eigenvectors using randomized methods,
    which is much faster than full eigendecomposition for large matrices.
    
    Args:
        A: Input symmetric matrix of shape (n, n)
        k: Number of eigenvalues/eigenvectors to compute
        n_oversamples: Additional random vectors for better accuracy
        n_iter: Number of power iterations for improved accuracy
        key: JAX random key (if None, will create one)
    
    Returns:
        eigenvalues: Top k eigenvalues in descending order, shape (k,)
        eigenvectors: Corresponding eigenvectors, shape (n, k)
    """
    if key is None:
        key = random.PRNGKey(42)
    
    n = A.shape[0]
    l = k + n_oversamples  # Total number of random vectors
    
    # Stage A: Randomized range finding
    # Generate random matrix
    key, subkey = random.split(key)
    Omega = random.normal(subkey, (n, l))
    
    # Initial matrix-vector products
    Y = A @ Omega
    
    # Power iterations for improved accuracy
    for _ in range(n_iter):
        Y = A @ Y
    
    # QR decomposition to get orthonormal basis
    Q, _ = jnp.linalg.qr(Y, mode='reduced')
    
    # Stage B: Direct eigendecomposition of smaller matrix
    # Project A onto the range of Q
    B = Q.T @ A @ Q  # This is the Rayleigh quotient
    
    # Compute eigendecomposition of the smaller matrix B
    eigenvals_B, eigenvecs_B = jnp.linalg.eigh(B)
    
    # Sort in descending order (largest eigenvalues first)
    idx = jnp.argsort(eigenvals_B)[::-1]
    eigenvals_B = eigenvals_B[idx]
    eigenvecs_B = eigenvecs_B[:, idx]
    
    # Recover the original eigenvectors
    eigenvectors = Q @ eigenvecs_B
    
    # Return only the first k components
    return eigenvals_B[:k], eigenvectors[:, :k]
```

```julia
import jax
import jax.numpy as jnp
from jax import random

def randomized_eigh(A, k, n_oversamples=10, n_iter=2, key=None, which='LM'):
    """
    Randomized eigendecomposition for symmetric/Hermitian matrices.
    
    Args:
        A: Input symmetric matrix (n x n)
        k: Number of eigenvalues/eigenvectors to compute
        n_oversamples: Additional samples for improved accuracy (default: 10)
        n_iter: Number of power iterations (default: 2)
        key: JAX PRNG key (if None, creates a new key)
        which: Selection criterion ('LM' for largest magnitude, 'LA' for largest algebraic)
    
    Returns:
        w: Array of top-k eigenvalues
        V: Matrix of corresponding eigenvectors (each column is an eigenvector)
    """
    if key is None:
        key = random.PRNGKey(0)
    
    n = A.shape[0]
    total_rank = min(k + n_oversamples, n)
    
    # Generate random test matrix
    key, subkey = random.split(key)
    dtype = A.dtype
    Omega = random.normal(subkey, (n, total_rank), dtype=dtype)
    
    # Power iterations
    Y = A @ Omega
    for _ in range(n_iter):
        # Orthonormalize to maintain stability
        Q = jnp.linalg.qr(Y, mode='reduced')[0]
        Y = A @ Q
    
    # Final orthonormal basis
    Q = jnp.linalg.qr(Y, mode='reduced')[0]
    
    # Form the small projected matrix
    B = Q.T @ A @ Q
    
    # Solve the small eigenvalue problem
    w_small, V_small = jnp.linalg.eigh(B)
    
    # Select eigenvalues/eigenvectors based on criterion
    if which == 'LM':  # Largest magnitude
        idx = jnp.argsort(jnp.abs(w_small))[::-1][:k]
    elif which == 'LA':  # Largest algebraic
        idx = jnp.argsort(w_small)[::-1][:k]
    else:
        raise ValueError("which must be 'LM' or 'LA'")
    
    w = w_small[idx]
    V_small = V_small[:, idx]
    
    # Project eigenvectors back to original space
    V = Q @ V_small
    
    return w, V
```

```julia
def randomized_eigh(A, k, n_oversamples=10, n_iter=2, key=None, which='LM'):
    """
    Randomized eigendecomposition for symmetric/Hermitian matrices.
    Supports both real symmetric and complex Hermitian matrices.
    
    Args:
        A: Input matrix (n x n), must be symmetric (real) or Hermitian (complex)
        k: Number of eigenvalues/eigenvectors to compute
        n_oversamples: Additional samples for improved accuracy (default: 10)
        n_iter: Number of power iterations (default: 2)
        key: JAX PRNG key (if None, creates a new key)
        which: Selection criterion ('LM' for largest magnitude, 'LA' for largest algebraic)
    
    Returns:
        w: Array of top-k eigenvalues
        V: Matrix of corresponding eigenvectors (each column is an eigenvector)
    """
    if key is None:
        key = random.PRNGKey(0)
    
    n = A.shape[0]
    total_rank = min(k + n_oversamples, n)
    is_complex = jnp.iscomplexobj(A)
    print(is_complex)
    
    # Generate random test matrix (complex if A is complex)
    key, subkey = random.split(key)
    dtype = A.dtype
    if is_complex:
        # Complex normal distribution with variance 1
        Omega_real = random.normal(subkey, (n, total_rank), dtype=dtype)
        key, subkey = random.split(key)
        Omega_imag = random.normal(subkey, (n, total_rank), dtype=dtype)
        Omega = Omega_real + 1j * Omega_imag
    else:
        Omega = random.normal(subkey, (n, total_rank), dtype=dtype)
    
    # Power iterations
    Y = A @ Omega
    for _ in range(n_iter):
        # Orthonormalize with modified Gram-Schmidt for stability
        Q, _ = jnp.linalg.qr(Y, mode='reduced')
        Y = A @ Q
    
    # Final orthonormal basis
    Q, _ = jnp.linalg.qr(Y, mode='reduced')
    
    # Form the small projected matrix (Q^H A Q for Hermitian matrices)
    if is_complex:
        B = jnp.conj(Q).T @ A @ Q  # Hermitian projection
    else:
        B = Q.T @ A @ Q           # Symmetric projection
    
    # Solve the small eigenvalue problem
    w_small, V_small = jnp.linalg.eigh(B)
    
    # Select eigenvalues/eigenvectors based on criterion
    if which == 'LM':  # Largest magnitude
        idx = jnp.argsort(jnp.abs(w_small))[::-1][:k]
    elif which == 'LA':  # Largest algebraic
        idx = jnp.argsort(w_small)[::-1][:k]
    else:
        raise ValueError("which must be 'LM' or 'LA'")
    
    w = w_small[idx]
    V_small = V_small[:, idx]
    
    # Project eigenvectors back to original space
    V = Q @ V_small
    
    return w, V
```

```julia
import jax
import jax.numpy as jnp
from jax import random
import jax.scipy as jsp

def randomized_eigh(A, k, n_oversamples=10, n_iter=2, key=None, which='LM'):
    """
    Randomized eigendecomposition for symmetric/Hermitian matrices.
    Supports both real symmetric and complex Hermitian matrices, including purely imaginary.
    
    Args:
        A: Input matrix (n x n), must be symmetric (real) or Hermitian (complex)
        k: Number of eigenvalues/eigenvectors to compute
        n_oversamples: Additional samples for improved accuracy (default: 10)
        n_iter: Number of power iterations (default: 2)
        key: JAX PRNG key (if None, creates a new key)
        which: Selection criterion ('LM' for largest magnitude, 'LA' for largest algebraic)
    
    Returns:
        w: Array of top-k eigenvalues
        V: Matrix of corresponding eigenvectors (each column is an eigenvector)
    """
    if key is None:
        key = random.PRNGKey(0)
    
    n = A.shape[0]
    total_rank = min(k + n_oversamples, n)
    is_complex = jnp.iscomplexobj(A)
    is_pure_imag = not jnp.isrealobj(A) and jnp.allclose(A, -jnp.conj(A))
    
    # Generate random test matrix
    key, subkey = random.split(key)
    dtype = A.dtype
    
    if is_pure_imag:
        # For purely imaginary matrices, generate real random matrix
        Omega = random.normal(subkey, (n, total_rank), dtype=jnp.real(A).dtype)
    elif is_complex:
        # Standard complex normal distribution
        key, subkey1, subkey2 = random.split(key, 3)
        Omega_real = random.normal(subkey1, (n, total_rank), dtype=jnp.real(A).dtype)
        Omega_imag = random.normal(subkey2, (n, total_rank), dtype=jnp.real(A).dtype)
        Omega = Omega_real + 1j * Omega_imag
    else:
        # Real case
        Omega = random.normal(subkey, (n, total_rank), dtype=dtype)
    
    # Power iterations with special handling for purely imaginary matrices
    Y = A @ Omega
    
    for _ in range(n_iter):
        # Handle QR decomposition for different matrix types
        if is_pure_imag:
            # Convert to complex for QR but keep imaginary nature
            Y_complex = Y.astype(jnp.complex64) if Y.dtype != jnp.complex128 else Y
            Q, _ = jnp.linalg.qr(Y_complex, mode='reduced')
        else:
            Q, _ = jnp.linalg.qr(Y, mode='reduced')
        
        # Apply matrix multiplication
        Y = A @ Q
    
    # Final orthonormal basis
    if is_pure_imag:
        Y_complex = Y.astype(jnp.complex64) if Y.dtype != jnp.complex128 else Y
        Q, _ = jnp.linalg.qr(Y_complex, mode='reduced')
    else:
        Q, _ = jnp.linalg.qr(Y, mode='reduced')
    
    # Form the small projected matrix with proper conjugation
    if is_complex or is_pure_imag:
        B = jnp.conj(Q).T @ (A @ Q)
    else:
        B = Q.T @ (A @ Q)
    
    # Solve the small eigenvalue problem
    w_small, V_small = jnp.linalg.eigh(B)
    
    # Select eigenvalues/eigenvectors based on criterion
    if which == 'LM':  # Largest magnitude
        idx = jnp.argsort(jnp.abs(w_small))[::-1][:k]
    elif which == 'LA':  # Largest algebraic
        idx = jnp.argsort(w_small)[::-1][:k]
    else:
        raise ValueError("which must be 'LM' or 'LA'")
    
    w = w_small[idx]
    V_small = V_small[:, idx]
    
    # Project eigenvectors back to original space
    V = Q @ V_small
    
    return w, V

# Test with purely imaginary Hermitian matrix
def test_pure_imaginary_hermitian():
    key = random.PRNGKey(42)
    
    # Create a purely imaginary Hermitian matrix (50 x 50)
    n = 50
    key, subkey = random.split(key)
    
    # Generate real skew-symmetric matrix (A = -A^T)
    A_real = random.normal(subkey, (n, n))
    A_skew = (A_real - A_real.T) / 2  # Skew-symmetric matrix
    
    # Convert to purely imaginary Hermitian matrix
    A = 1j * A_skew
    
    # Verify Hermitian property: A should equal A^H
    hermitian_check = jnp.allclose(A, jnp.conj(A).T)
    print(f"Matrix is Hermitian: {hermitian_check}")
    
    # Compute eigenvalues
    w, V = randomized_eigh(A, k=50, n_oversamples=15, n_iter=3, key=key, which='LM')
    
    # Verify eigenproperties
    residuals = A @ V - V * w
    residual_norm = jnp.linalg.norm(residuals, 'fro')
    print(f"Residual norm: {residual_norm:.4e}")
    
    # Verify orthogonality
    ortho_test = jnp.conj(V).T @ V
    eye = jnp.eye(V.shape[1])
    ortho_error = jnp.linalg.norm(ortho_test - eye)
    print(f"Orthogonality error: {ortho_error:.4e}")
    
    # Compare with direct method
    w_full, V_full = jnp.linalg.eigh(A)  # Uses Hermitian algorithm
    idx_full = jnp.argsort(jnp.abs(w_full))[::-1]
    w_full_sorted = w_full[idx_full]
    # w_full_sorted = jnp.sort(w_full)[::-1]  # Top eigenvalues
    # print("Direct V:", V_full[:10,:10])
    # print("Randomized V:", V[:10,:10])

    
    print("\nTop 10 eigenvalues comparison:")
    print("Randomized:", w)
    print("Direct:", w_full_sorted)
    print(f"Eigenvalue error: {jnp.linalg.norm(jnp.sort(w) - jnp.sort(w_full_sorted))/jnp.linalg.norm(jnp.sort(w_full_sorted)):.4e}")

if __name__ == "__main__":
    test_pure_imaginary_hermitian()
```

```julia
def randomized_svd(A, k, n_oversamples=10, n_iter=2, key=None):
    """
    Randomized SVD algorithm for truncated SVD approximation.
    
    Args:
        A: Input matrix (m x n)
        k: Target rank (number of singular values/vectors to compute)
        n_oversamples: Additional samples for improved accuracy (default: 10)
        n_iter: Number of power iterations (default: 2)
        key: JAX PRNG key (if None, creates a new key)
    
    Returns:
        U: m x k matrix of left singular vectors
        s: length-k vector of singular values
        Vh: k x n matrix of right singular vectors (as rows)
    """
    if key is None:
        key = random.PRNGKey(0)
    
    m, n = A.shape
    l = n_oversamples
    total_rank = min(k + l, min(m, n))
    
    # Generate random test matrix
    key, subkey = random.split(key)
    dtype = A.dtype
    Omega = random.normal(subkey, (n, total_rank), dtype=dtype)
    
    # Power iterations to improve approximation quality
    Y = A @ Omega
    for _ in range(n_iter):
        Y = jnp.linalg.qr(Y, mode='reduced')[0]
        Z = jnp.linalg.qr(Y.T @ A, mode='reduced')[0].T
        Y = A @ Z
    
    # Orthonormal basis for the range
    Q = jnp.linalg.qr(Y, mode='reduced')[0]
    
    # Project matrix to smaller space
    B = jnp.conj(Q).T @ A
    
    # SVD of the projected matrix
    U_tilde, s, Vh = jnp.linalg.svd(B, full_matrices=False)
    
    # Truncate to target rank
    U_tilde = U_tilde[:, :k]
    s = s[:k]
    Vh = Vh[:k, :]
    
    # Form full left singular vectors
    U = Q @ U_tilde
    
    return U, s, Vh

# Usage example
if __name__ == "__main__":
    # Set up JAX PRNG key
    key = random.PRNGKey(42)
    
    # Create a sample matrix (3000 x 200)
    key, subkey = random.split(key)
    A = random.normal(subkey, (3000, 200))
    
    # Compute randomized SVD
    U, s, Vh = randomized_svd(A, k=10, n_oversamples=20, n_iter=2, key=key)
    
    # Verify reconstruction error
    reconstruction = U @ jnp.diag(s) @ Vh
    error = jnp.linalg.norm(A - reconstruction, ord='fro')
    print(f"Reconstruction error: {error:.4e}")
    
    # Compare with full SVD (for small matrices)
    if A.shape[0] < 1000 and A.shape[1] < 1000:
        U_full, s_full, Vh_full = jnp.linalg.svd(A, full_matrices=False)
        s_full = s_full[:10]
        print(f"Top 10 singular values error: {jnp.linalg.norm(s - s_full):.4e}")
```

```julia
import jax
import jax.numpy as jnp
from jax import random, lax
from jax.scipy.linalg import svd

def adaptive_randomized_eigh(A, tol=1e-6, max_rank=200, block_size=10, 
                            n_iter=2, key=None, verbose=False):
    """
    Adaptive randomized eigendecomposition for symmetric matrices.
    
    Iteratively increases the rank until the residual norm is below tolerance.
    
    Args:
        A: Symmetric input matrix (n x n)
        tol: Tolerance for residual norm (default: 1e-6)
        max_rank: Maximum rank to compute (default: 100)
        block_size: Number of vectors to add per iteration (default: 10)
        n_iter: Number of power iterations per step (default: 2)
        key: JAX PRNG key (if None, creates new key)
        verbose: Print progress information (default: False)
        
    Returns:
        w: Array of eigenvalues (in descending order)
        V: Matrix of corresponding eigenvectors
        residual_norm: Final residual norm
        total_rank: Final rank used
    """
    if key is None:
        key = random.PRNGKey(0)
    
    n = A.shape[0]
    dtype = A.dtype
    min_rank = min(block_size, max_rank)
    
    # Initialize basis and projection matrix
    key, subkey = random.split(key)
    Q = random.normal(subkey, (n, min_rank), dtype=dtype)
    total_rank = min_rank
    
    # Define helper functions
    def qr_safe(X):
        """Safe QR decomposition with reduced mode."""
        Q, R = jnp.linalg.qr(X, mode='reduced')
        return Q, R
    
    def power_iteration(Q, A, n_iter):
        """Perform power iterations to improve basis quality."""
        def body_fn(_, Qi):
            Qi = A @ Qi
            Qi, _ = qr_safe(Qi)
            return Qi
        return lax.fori_loop(0, n_iter, body_fn, Q)
    
    def update_basis(Q, A, total_rank, block_size, key, n_iter):
        """Extend the basis with new vectors and apply power iterations."""
        # Generate new random vectors
        key, subkey = random.split(key)
        Omega = random.normal(subkey, (n, block_size), dtype=dtype)
        
        # Orthogonalize against current basis
        if total_rank > 0:
            Omega = Omega - Q @ (Q.T @ Omega)
        
        # Orthonormalize new vectors
        Q_new, _ = qr_safe(Omega)
        actual_block_size = Q_new.shape[1]
        
        # Combine bases
        Q_ext = jnp.hstack([Q, Q_new]) if total_rank > 0 else Q_new
        total_rank_ext = total_rank + actual_block_size
        
        # Apply power iterations to the extended basis
        Q_ext = power_iteration(Q_ext, A, n_iter)
        return Q_ext, total_rank_ext, key, actual_block_size
    
    def compute_eigenproblem(Q, A):
        """Compute the projected eigenproblem."""
        B = Q.T @ (A @ Q)
        w, V_small = jnp.linalg.eigh(B)
        # Sort in descending order
        idx = jnp.argsort(w)[::-1]
        w = w[idx]
        V_small = V_small[:, idx]
        V_full = Q @ V_small
        return w, V_full
    
    def compute_residual(A, V, w):
        """Compute residual norm."""
        R = A @ V - V * w
        return jnp.linalg.norm(R, 'fro')
    
    # Main iteration loop
    residual_norm = jnp.inf
    iteration = 0
    converged = False
    
    while not converged and total_rank < max_rank:
        iteration += 1
        
        # Extend basis and apply power iterations
        Q, total_rank, key, actual_block_size = update_basis(
            Q, A, total_rank, block_size, key, n_iter
        )
        
        # Compute current eigenproblem
        w, V = compute_eigenproblem(Q, A)
        
        # Compute residual norm
        residual_norm = compute_residual(A, V, w)
        
        # Check convergence
        converged = residual_norm < tol
        
        if verbose:
            print(f"Iter {iteration}: Rank {total_rank}, Residual {residual_norm:.4e}")
        
        # Break if we've reached max rank
        if total_rank >= max_rank:
            if verbose:
                print(f"Reached max rank {max_rank}")
            break
    
    # Final computation with converged basis
    w, V = compute_eigenproblem(Q, A)
    residual_norm = compute_residual(A, V, w)
    
    if verbose:
        print(f"Final rank: {total_rank}, Final residual: {residual_norm:.4e}")
    
    return w, V, residual_norm, total_rank
```
